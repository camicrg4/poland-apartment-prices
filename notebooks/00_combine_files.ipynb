# Combining Monthly Files
Combines all monthly CSV files from Kaggle into one file for sales and one for rentals, adding a new `snapshot_month` column.

```python
import pandas as pd
import glob
import os
```

```python
files_sales = sorted(glob.glob('../data/raw/apartments_pl_*.csv'))
print(len(files_sales))
files_sales
```

```python
def read_with_month(path):
    df = pd.read_csv(path)
    name = os.path.basename(path)
    df['snapshot_month'] = name[-11:-4].replace('_', '-')
    return df
```

```python
sales = pd.concat([read_with_month(f) for f in files_sales], ignore_index=True)
print(sales.shape)
sales.head()
```

```python
sales['snapshot_month'].value_counts().sort_index()
```

```python
files_rent = sorted(glob.glob('../data/raw/apartments_rent_pl_*.csv'))
print(len(files_rent))   # trebuie sa fie 8

rent = pd.concat([read_with_month(f) for f in files_rent], ignore_index=True)
print(rent.shape)
rent['snapshot_month'].value_counts().sort_index()
```

```python
sales.to_csv('../data/processed/sales_all.csv', index=False)
rent.to_csv('../data/processed/rent_all.csv', index=False)
print('Salvat:', sales.shape, rent.shape)
```
